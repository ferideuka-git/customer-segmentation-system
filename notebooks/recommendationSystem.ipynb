```python
def recommend_products(product_code, n=5):
    if product_code not in item_similarity_df.index:
        return pd.DataFrame(columns=["StockCode", "Similarity", "Description"])
    similarities=item_similarity_df[product_code]
    similarities = similarities.drop(product_code)
    top_products = similarities.sort_values(
    ascending=False
    ).head(n)
    recommendations = top_products.reset_index()
    recommendations.columns = ["StockCode", "Similarity"]
    recommendations = recommendations.merge(
        product_info,
        on="StockCode",
        how="left"
    )
    return recommendations
```

```python
evaluation_results = []

for customer_id in common_customers:

    customer_products = get_customer_products(customer_id)

    customer_recommendations = {}

    for product in customer_products:

        if product in train_item_similarity_df.index:

            similar_products = (
                train_item_similarity_df[product]
                .drop(product)
                .sort_values(ascending=False)
                .head(5)
            )

            customer_recommendations[product] = similar_products

    recommendations_list = []

    for source_product, recs in customer_recommendations.items():

        temp = recs.reset_index()

        temp.columns = ["StockCode", "Similarity"]

        temp["SourceProduct"] = source_product

        recommendations_list.append(temp)

    if not recommendations_list:
        continue

    all_recommendations_df = pd.concat(
        recommendations_list,
        ignore_index=True
    )

    bought_products = set(customer_products)

    filtered_recommendations = all_recommendations_df[
        ~all_recommendations_df["StockCode"].isin(bought_products)
    ]

    recommendations_unique = (
        filtered_recommendations
        .sort_values("Similarity", ascending=False)
        .drop_duplicates("StockCode")
    )

    top_5Unique = recommendations_unique.head(5)

    test_products = test_df[
        test_df["CustomerID"] == customer_id
    ]["StockCode"].unique().tolist()

    recommended_products = top_5Unique["StockCode"].tolist()

    hits = set(recommended_products) & set(test_products)

    evaluation_results.append({
        "CustomerID": customer_id,
        "Recommended": recommended_products,
        "TestProducts": test_products,
        "Hits": list(hits),
        "HitCount": len(hits)
    })
```

```python
evaluation_df = pd.DataFrame(evaluation_results)

evaluation_df.head()
```

```python
total_customers = len(evaluation_df)

customers_with_hits = (
    evaluation_df["HitCount"] > 0
).sum()

hit_rate = customers_with_hits / total_customers

print("Total customers evaluated:", total_customers)
print("Customers with at least one hit:", customers_with_hits)
print("Hit Rate:", hit_rate)
```

```python
precision_at_5 = (
    evaluation_df["HitCount"] / 5
).mean()

print("Precision@5:", precision_at_5)
```

```python
evaluation_df["RecallAt5"] = (
    evaluation_df["HitCount"] /
    evaluation_df["TestProducts"].apply(len)
)

recall_at_5 = evaluation_df["RecallAt5"].mean()

print("Recall@5:", recall_at_5)
```
